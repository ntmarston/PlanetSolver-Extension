```python
from astropy import units as u
import numpy as np
from astropy.constants import G, sigma_sb, k_B
```

$\Delta S = \frac{\Delta Q}{T}$, where $T$ is approximated as $T_{eq}$ and core heating is ignored

- Core cooling might prolong the inflation?

$E_{imp} = \eta M_{imp} \frac{GM_p}{R_c}$

$S-S_0 = \frac{E_{imp}}{T_{eq}}$

```python
#Input parameters
#Planet

Impact_mass_fraction = 0.005
M_p = 8 * u.Mearth
R_p = 10 * u.Rearth
f_env = 0.2
#Star
L = 1 * u.Lsun

a = 0.3 * u.AU
#Backend assumptions
eta = 0.5 #Check paper to verify realistic values, decide what upper limit should be
mmw = 28.5 * u.u # From titan's atmosphere, placeholder until more reading is done 1981Natur.292..688S
A_b = 0.3


M_core = (1-f_env) * M_p
M_imp = Impact_mass_fraction * M_core
R_c = R_p * (M_core/M_p) ** 4  
E_imp = eta * M_imp * (G * M_p)/R_c
E_imp = E_imp.si
E_imp
```

```python
T_eq = (L*(1-A_b)/(16 * sigma_sb * np.pi * a**2))**(1/4)
```

```python
entropy = E_imp/T_eq
entropy.si
```

```python
M_atm = M_p - M_core

N_bary = M_atm / mmw
N_bary = N_bary.si
N_bary
```

```python
6.0 * N_bary * k_B
```

```python
(entropy/N_bary/k_B).si
```
