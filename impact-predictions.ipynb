```python
#Input parameters
#Planet

Impact_mass_fraction = 0.015 # <-- Free variable
M_p = 8 * u.Mearth
R_p = 10 * u.Rearth
f_env = 0.2 # Do parameter sweep
Z = 30 #atmospheric entropy, times solar

#Test param
current_spec_entropy = 6.0 #kB/Baryon

#Star
L = 1 * u.Lsun # Checked

a = 0.3 * u.AU # Checked

#Backend assumptions

eta = 0.5 # Realistic ish value
hhe_mmw = 2.247 * u.u # Taken from PlanetSolver/HheComp.cpp and consistent with Biersteker paper
metals_mmw = 16.93 * u.u #Taken from PlanetSolver, assumes molecules aren't forming
A_b = 0.3 # Probably fine, but double check


#Calc planet parameters
M_core = (1-f_env) * M_p
M_imp = Impact_mass_fraction * M_core
R_c = R_p * (M_core/M_p) ** 4  
M_atm = M_p - M_core

#Calc MMW(Z)
Nfrac_metal = Z*0.00276 # Fraction of the total number of molecules heavier than helium
mmw = (1-Nfrac_metal)*hhe_mmw + (Nfrac_metal)*metals_mmw

N_molecules = ( M_atm / mmw )
N_bary = N_molecules * 2.3 # Mean molecular weight is essentially just the number of baryons per molecule as well?
N_bary = N_bary.si #Multiply by number of baryons per molecule
N_bary

S0 = current_spec_entropy * N_bary * k_B #Total entropy of atmosphere pre-impact

T_eq = (L*(1-A_b)/(16 * sigma_sb * np.pi * a**2))**(1/4) #Where did this come from and can we verify and cite plz

T_core = 2500 * u.K #Take from planetsolver result? All the temp values up to the core boundary are equal so the mode of the column will probably give the right value
```

```python
#Calculate energy of impact based on input
E_imp = eta * M_imp * (G * M_p)/R_c
E_imp = E_imp.si
E_imp
```

```python
#Calculate entropy to the system added according to impact energy
Delta_Entropy = (E_imp/T_eq).si
Delta_Entropy
```

```python
Delta_spec_entropy = (Delta_Entropy/N_bary/k_B).si #Convert to specific entropy
Delta_spec_entropy
```

```python
# Core energy
Cvc = 7.5e6 * u.erg * u.g**(-1) * u.K**(-1) # Taken from Biersteker eq. 9

E_core = (Cvc * M_core * T_core).si
```
